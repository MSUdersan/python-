```python
def add(*args):
    total = 0
    for num in args:
        total += num
    return total
print(add(1,2,7))
```

```python
def create_profile(**kwargs):
    print("user profile")
    
    for key,value in kwargs.items():
        print(f"{key}:{value}") 
        
create_profile(name ="san",age=24)
```
